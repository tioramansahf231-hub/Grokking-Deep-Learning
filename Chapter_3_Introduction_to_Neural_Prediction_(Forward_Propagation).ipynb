# Chapter 3: Introduction to Neural Prediction (Forward Propagation)

## Overview & Teori Dasar
Forward propagation adalah proses bagaimana neural network mengubah data input menjadi sebuah prediksi. Pada bab ini, dipelajari 4 arsitektur dasar forward propagation:
1. **Single Input & Output:** Menghitung prediksi dengan mengalikan satu input dengan satu bobot (weight).
2. **Multiple Inputs, Single Output:** Menggabungkan beberapa input menggunakan perkalian titik (*dot product*) dengan vektor bobot.
3. **Single Input, Multiple Outputs:** Memprediksi beberapa target dari satu input dengan mengalikan input ke setiap elemen vektor bobot.
4. **Multiple Inputs & Outputs (Stacked Neural Network):** Menggunakan perkalian matriks untuk menghubungkan beberapa input ke beberapa output melalui lapisan tersembunyi (*hidden layer*).

---

## 1. Predict with Single Input & Output

$$\text{prediction} = \text{input} \times \text{weight}$$

```python
import numpy as np

# Inisialisasi bobot sederhana
weight = 0.1

def neural_network(input, weight):
    prediction = input * weight
    return prediction

# Contoh input: jumlah jari-jari (number of toes)
number_of_toes = [8.5, 9.5, 9.9, 9.0]
input_data = number_of_toes[0]

pred = neural_network(input_data, weight)
print("Hasil Prediksi Single Input:", pred)

Penjelasan Teoretis:

1.Input: Merupakan variabel independen dari dunia nyata.

2.Weight: Mengukur seberapa besar pengaruh input terhadap output.

3.Prediction: Hasil perkalian linier antara input dan bobot.

2. Predict with Multiple Inputs$$\text{prediction} = \sum_{i} (x_i \cdot w_i) = \vec{x} \cdot \vec{w}$$

````python
import numpy as np

# Bobot untuk masing-masing fitur: [toes, wl_ratio, nfans]
weights = np.array([0.1, 0.2, 0])

def neural_network_multi_input(inputs, weights):
    # Perkalian titik (dot product) antara vektor input dan bobot
    prediction = inputs.dot(weights)
    return prediction

toes = [8.5, 9.5, 9.9, 9.0]
wl_ratio = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input_data = np.array([toes[0], wl_ratio[0], nfans[0]])

pred = neural_network_multi_input(input_data, weights)
print("Hasil Prediksi Multiple Inputs:", pred)
````

Penjelasan Teoretis:

Dot Product: Mengalikan elemen-elemen yang seposisi pada dua vektor, lalu menjumlahkan seluruh hasilnya. Ini merepresentasikan kombinasi linier dari seluruh fitur input.

3. Predict with Multiple Outputs

````python
weights = np.array([0.3, 0.2, 0.9])

def neural_network_multi_output(input_data, weights):
    # Mengalikan skalar input ke setiap elemen vektor bobot
    prediction = input_data * weights
    return prediction

wl_ratio = [0.65, 0.8, 0.8, 0.9]
input_data = wl_ratio[0]

pred = neural_network_multi_output(input_data, weights)
print("Hasil Prediksi Multiple Outputs:", pred)
````

Penjelasan Teoretis:

Satu sinyal masukan dipancarkan (element-wise multiplication) ke beberapa bobot untuk menghasilkan beberapa variabel prediksi secara independen.

4. Multiple Inputs & Multiple Outputs (Deep Network)$$\text{hidden} = \text{input} \cdot W_{ih}$$$$\text{prediction} = \text{hidden} \cdot W_{hp}$$

````python
# Bobot Input-to-Hidden Layer
ih_wgt = np.array([
    [0.1, 0.2, -0.1],
    [-0.1, 0.1, 0.9],
    [0.1, 0.4, 0.1]
])

# Bobot Hidden-to-Prediction Layer
hp_wgt = np.array([
    [0.3, 1.1, -0.3],
    [0.1, 0.2, 0.0],
    [0.0, 1.3, 0.1]
])

weights = [ih_wgt, hp_wgt]

def neural_network_deep(inputs, weights):
    # Layer 1: Hidden Layer
    hidden = inputs.dot(weights[0])
    # Layer 2: Output Layer
    prediction = hidden.dot(weights[1])
    return prediction

input_data = np.array([toes[0], wl_ratio[0], nfans[0]])

pred = neural_network_deep(input_data, weights)
print("Hasil Prediksi Stacked Neural Network:", pred)
````

Penjelasan Teoretis:

Hidden Layer: Lapisan perantara yang memungkinkan jaringan mempelajari representasi tingkat tinggi dari kombinasi data input.

Deep Neural Network: Menumpuk beberapa matriks bobot (stacked weights) memungkinkan pembuatan model keputusan yang lebih kompleks.
