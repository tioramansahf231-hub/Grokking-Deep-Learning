Overview & Teori Dasar
Pada bab penutup ini, dipelajari dasar-dasar Deep Reinforcement Learning (DRL), yaitu cabang machine learning di mana agen belajar mengambil keputusan optimal melalui interaksi langsung dengan lingkungan (environment):

Agent & Environment: Agen bertindak berdasarkan state lingkungan, lalu menerima imbalan (reward) atau hukuman (penalty).

Policy: Strategi atau pemetaan dari state lingkungan ke aksi yang akan diambil oleh agen.

Q-Learning / Policy Gradient: Menggunakan neural network untuk memprediksi reward masa depan berdasarkan aksi yang dipilih.

1. Simulasi Sederhana Environment dan Policy Gradient

```python
import numpy as np

# Set seed untuk konsistensi
np.random.seed(1)

# Inisialisasi Bobot Agen (3 State -> 2 Action)
weights = 2 * np.random.random((3, 2)) - 1
alpha = 0.1

# Lingkungan Sederhana (3 fitur state)
state = np.array([1, 0, 1])

# Fungsi Softmax untuk memilih aksi (Policy)
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

# Loop Pelatihan Agen (10 Episode)
for episode in range(10):
    # 1. Agen Mengamati State & Memilih Aksi (Forward Pass)
    probabilities = softmax(np.dot(state, weights))

    # Pilih aksi berdasarkan distribusi probabilitas
    action = np.random.choice([0, 1], p=probabilities)

    # 2. Lingkungan Memberikan Reward (Misal: Aksi 1 memberi reward positif)
    reward = 1.0 if action == 1 else -1.0

    # 3. Hitung Synthetic Target / Error berdasarkan Reward
    target = np.zeros(2)
    target[action] = 1.0 if reward > 0 else 0.0

    # Backpropagation untuk Memperbarui Policy Agen
    delta = probabilities - target
    weight_delta = np.outer(state, delta)
    weights -= alpha * weight_delta

    print(f"Episode {episode+1:2d} | Aksi Terpilih: {action} | Reward: {reward:+.1f} | Probabilitas Aksi [0, 1]: {probabilities.round(3)}")
```

Penjelasan Teoretis:

Trial and Error Learning: Agen tidak diberi tahu aksi mana yang paling tepat secara langsung, melainkan harus mengeksplorasi pilihan aksi dan belajar dari reward yang diperoleh.

Credit Assignment Problem: Menilai seberapa besar kontribusi aksi tertentu di masa lalu terhadap reward yang diterima di masa depan.
