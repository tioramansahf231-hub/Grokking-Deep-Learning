Overview & Teori DasarPada bab ini, dipelajari berbagai Fungsi Aktivasi non-linier dan peran pentingnya dalam memodelkan probabilitas:Sigmoid: Memetakan nilai input apa pun ke dalam rentang rentang probabilitas $0$ hingga $1$.Tanh (Hyperbolic Tangent): Memetakan input ke dalam rentang $-1$ hingga $1$, sangat baik untuk hidden layer karena memusatkan data di sekitar angka nol (zero-centered).Softmax: Digunakan pada output layer untuk klasifikasi multi-kelas, mengubah seluruh skor masukan menjadi distribusi probabilitas total bernilai $1$ ($100\%$).

1. Fungsi Aktivasi Sigmoid & Tanh beserta Turunannya

```python
import numpy as np

# Fungsi Sigmoid & Turunannya
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid2deriv(output):
    return output * (1 - output)

# Fungsi Tanh & Turunannya
def tanh(x):
    return np.tanh(x)

def tanh2deriv(output):
    return 1 - (output ** 2)

# Contoh Pengujian Nilai Input
x = np.array([-2.0, 0.0, 2.0])

print("Sigmoid Output:", sigmoid(x))
print("Sigmoid Derivative:", sigmoid2deriv(sigmoid(x)))
print("Tanh Output:", tanh(x))
print("Tanh Derivative:", tanh2deriv(tanh(x)))
```

Penjelasan Teoretis:

Sigmoid: Sangat cocok untuk masalah klasifikasi biner (Output Layer).

Tanh: Merupakan bentuk Sigmoid yang digeser. Karena bernilai negatif dan positif, Tanh mencegah pembiasan arah gradien pada hidden layer.

2. Fungsi Aktivasi Softmax untuk Distribusi ProbabilitasSoftmax menghitung eksponensial dari setiap nilai input lalu membaginya dengan jumlah total eksponensial seluruh elemen.$$\text{Softmax}(x_i) = \frac{e^{x_i}}{\sum_{j} e^{x_j}}$$

```python
import numpy as np

def softmax(x):
    # Kurangi dengan nilai maksimum untuk stabilitas numerik
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

# Contoh skor mentah (logits) dari 3 kelas
logits = np.array([2.0, 1.0, 0.1])
probabilities = softmax(logits)

print("Skor Mentah (Logits):", logits)
print("Distribusi Probabilitas (Softmax):", probabilities)
print("Total Probabilitas:", np.sum(probabilities))
```

Penjelasan Teoretis:Softmax: Mengubah angka mentah (logits) menjadi persentase probabilitas. Total jumlah nilai seluruh kelas hasil Softmax pasti tepat bernilai $1$.

3. Pelatihan Sederhana Menggunakan Sigmoid pada Dataset MNIST

```python
import numpy as np

np.random.seed(1)

# Input dummy (1 sampel x 4 fitur) dan Target Biner
input_data = np.array([[0.5, -0.2, 0.1, 0.8]])
target = np.array([[1]])
alpha = 0.5

# Bobot awal
weights_0_1 = 2 * np.random.random((4, 3)) - 1
weights_1_2 = 2 * np.random.random((3, 1)) - 1

for iteration in range(50):
    # Forward Pass dengan Sigmoid
    layer_1 = tanh(np.dot(input_data, weights_0_1))
    layer_2 = sigmoid(np.dot(layer_1, weights_1_2))

    # Error & Delta
    error = np.sum((layer_2 - target) ** 2)
    layer_2_delta = (layer_2 - target) * sigmoid2deriv(layer_2)

    # Backpropagation
    layer_1_delta = layer_2_delta.dot(weights_1_2.T) * tanh2deriv(layer_1)

    # Update Bobot
    weights_1_2 -= alpha * layer_1.T.dot(layer_2_delta)
    weights_0_1 -= alpha * input_data.T.dot(layer_1_delta)

    if iteration % 10 == 9:
        print(f"Iterasi {iteration+1:2d} | Error: {error:.6f} | Prediksi: {layer_2[0][0]:.4f}")
```
