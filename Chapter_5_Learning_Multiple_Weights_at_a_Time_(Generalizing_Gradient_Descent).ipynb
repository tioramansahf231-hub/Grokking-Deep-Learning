Pada bab ini, kita memperluas penggunaan Gradient Descent untuk menyesuaikan banyak bobot secara bersamaan. Proses pembelajaran melibatkan:Multiple Inputs & Single Output: Mengupdate vektor bobot berdasarkan beberapa sinyal input murni.Single Input & Multiple Outputs: Membagikan satu sinyal kesalahan (delta) ke banyak bobot target.Multiple Inputs & Multiple Outputs: Memperbarui matriks bobot (weight matrix) menggunakan perkalian luar (outer product).1. Gradient Descent dengan Banyak Input (Multiple Inputs)Ketika ada banyak input, setiap bobot harus diperbarui secara independen berdasarkan kontribusi inputnya terhadap total error.$$\text{weight_delta}_i = text{delta} times text{input}_i$$

```python
import numpy as np

# Inisialisasi bobot awal untuk 3 fitur
weights = np.array([0.1, 0.2, -0.1])

# Data input [toes, wl_ratio, nfans] dan target
toes = [8.5, 9.5, 9.9, 9.0]
wl_ratio = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

win_or_lose_binary = [1, 1, 0, 1]

# Ambil sampel pertama
inp = np.array([toes[0], wl_ratio[0], nfans[0]])
target = win_or_lose_binary[0]
alpha = 0.01

def neural_network(inputs, weights):
    return inputs.dot(weights)

for iteration in range(20):
    # 1. PREDICT
    pred = neural_network(inp, weights)

    # 2. COMPARE
    error = (pred - target) ** 2
    delta = pred - target

    # 3. LEARN (Hitung weight_delta untuk setiap bobot)
    weight_deltas = inp * delta

    # Update setiap bobot menggunakan alpha
    weights -= alpha * weight_deltas

    print(f"Iterasi {iteration+1:2d} | Error: {error:.6f} | Prediksi: {pred:.6f}")
```

Penjelasan Teoretis:

Weight Deltas: Hasil perkalian antara delta (selisih prediksi dan target) dengan masing-masing input.

Input yang bernilai 0 tidak akan mengubah bobot terkait karena tidak berkontribusi pada kesalahan prediksi.

2. Gradient Descent dengan Banyak Output (Multiple Outputs)
Jika satu input memprediksi banyak output, satu nilai input dikalikan ke vektor delta untuk memperbarui seluruh vektor bobot.

```python
# Satu input memprediksi 3 target berbeda [hurt?, win?, sad?]
weights = np.array([0.3, 0.2, 0.9])
wl_ratio_single = wl_ratio[0]
targets = np.array([1, 1, 0])
alpha = 0.1

for iteration in range(20):
    # 1. PREDICT
    pred = wl_ratio_single * weights

    # 2. COMPARE
    error = (pred - targets) ** 2
    deltas = pred - targets

    # 3. LEARN
    weight_deltas = wl_ratio_single * deltas
    weights -= alpha * weight_deltas

    print(f"Iterasi {iteration+1:2d} | Total Error: {np.sum(error):.6f}")
```

Penjelasan Teoretis:

Deltas Vector: Menghitung error independen untuk setiap variabel output.

Setiap elemen bobot disesuaikan secara terpisah agar masing-masing output mendekati targetnya.

3. Gradient Descent dengan Banyak Input & Banyak OutputProses ini menggunakan operasi Outer Product antara vektor input dan vektor delta untuk memperbarui matriks bobot secara bersamaan.$$\text{weight_deltas} = \text{outer_product}(\text{deltas}, \text{inputs})$$

```python
# Matriks bobot berukuran (3 output x 3 input)
weights = np.array([
    [0.1, 0.1, -0.3], # Hurt?
    [0.1, 0.2, 0.0],  # Win?
    [0.0, 1.3, 0.1]   # Sad?
])

inputs = np.array([toes[0], wl_ratio[0], nfans[0]])
targets = np.array([1, 1, 0])
alpha = 0.01

for iteration in range(20):
    # 1. PREDICT
    pred = weights.dot(inputs)

    # 2. COMPARE
    error = (pred - targets) ** 2
    deltas = pred - targets

    # 3. LEARN (Outer product untuk menghasilkan matriks weight_deltas)
    weight_deltas = np.outer(deltas, inputs)

    # Update matriks bobot
    weights -= alpha * weight_deltas

    print(f"Iterasi {iteration+1:2d} | Total Error: {np.sum(error):.6f}")
```

Penjelasan Teoretis:

Outer Product (np.outer): Mengalikan setiap elemen dari vektor deltas dengan setiap elemen dari vektor inputs untuk membentuk matriks perubahan bobot berukuran sama dengan matriks weights.
