Overview & Teori Dasar
Pada Chapter 3, kita hanya membuat prediksi (forward propagation). Pada bab ini, kita belajar bagaimana neural network dapat belajar dari kesalahannya melalui siklus Predict, Compare, Learn:

Predict: Menghitung nilai prediksi berdasarkan input dan bobot saat ini.

Compare: Menghitung seberapa jauh prediksi dari target sebenarnya (Mean Squared Error).

Learn: Mengubah nilai bobot untuk mengurangi tingkat kesalahan (error) menggunakan algoritma Gradient Descent.

1. Hot and Cold Learning (Metode Trial & Error Sederhana)
Metode paling sederhana untuk mengurangi error adalah dengan mencoba menambah atau mengurang bobot secara manual, lalu melihat arah mana yang menurunkan error.



```python
# Inisialisasi input, target, dan bobot
weight = 0.5
input_data = 0.5
goal_prediction = 0.8

# Langkah perubahan bobot (step size)
step_amount = 0.001

for iteration in range(1101):
    # 1. PREDICT
    prediction = input_data * weight

    # 2. COMPARE (Pure Error & Mean Squared Error)
    error = (prediction - goal_prediction) ** 2

    print(f"Iterasi {iteration} - Error: {error:.5f} | Prediksi: {prediction:.5f}")

    # 3. LEARN (Hot and Cold)
    # Coba jika bobot dinaikkan
    up_prediction = input_data * (weight + step_amount)
    up_error = (up_prediction - goal_prediction) ** 2

    # Coba jika bobot diturunkan
    down_prediction = input_data * (weight - step_amount)
    down_error = (down_prediction - goal_prediction) ** 2

    # Perbarui bobot ke arah yang menghasilkan error lebih kecil
    if down_error < up_error:
        weight = weight - step_amount
    if down_error > up_error:
        weight = weight + step_amount
```

Penjelasan Teoretis:

Hot and Cold Learning: Mencoba menggeser bobot ke atas dan ke bawah untuk melihat opsi mana yang menghasilkan error lebih rendah.

Kekurangan: Efisien hanya untuk kasus sangat sederhana, tetapi lambat dan tidak efisien jika memiliki banyak bobot.

2. Calculating Direction and Amount (Gradient Descent)Dibandingkan mencoba-coba seperti Hot and Cold, Gradient Descent menggunakan kalkulus (turunan/derivative) untuk menghitung arah dan besaran perubahan bobot secara langsung.$$\text{direction_and_amount} = (text{prediction} - text{goal\_prediction}) \times \text{input}$$$$\text{weight} = \text{weight} - \text{direction_and_amount}$$

```python
weight = 0.5
goal_pred = 0.8
input_data = 0.5

for iteration in range(20):
    # 1. PREDICT
    pred = input_data * weight

    # 2. COMPARE (MSE)
    error = (pred - goal_pred) ** 2

    # 3. LEARN (Calculated Delta & Weight Delta)
    delta = pred - goal_pred            # Mengukur seberapa besar dan ke arah mana error terjadi
    weight_delta = input_data * delta   # Menghitung penyesuaian bobot berdasarkan input

    # Perbarui bobot
    weight -= weight_delta

    print(f"Iterasi {iteration+1} - Error: {error:.6f} | Prediksi: {pred:.6f} | Weight: {weight:.6f}")
```

Penjelasan Teoretis:Delta ($\text{pred} - \text{goal_pred}$): Selisih murni antara prediksi dan target.Weight Delta ($\text{input} \times \text{delta}$): Menghitung seberapa banyak bobot harus diubah. Jika input bernilai besar, perubahan pada bobot juga akan lebih besar.

3. Mencegah Over shooting dengan Alpha (Learning Rate)Jika nilai input terlalu besar, perubahan bobot bisa menjadi sangat drastis sehingga prediksi malah menjauh dari target (divergence / overshooting). Untuk mengontrolnya, digunakan parameter Alpha ($\alpha$) atau Learning Rate.$$\text{weight_delta} = \alpha \times (\text{delta} \times \text{input})$$

```python
weight = 0.5
goal_pred = 0.8
input_data = 2.0
alpha = 0.1  # Learning rate untuk mengontrol kecepatan belajar

for iteration in range(20):
    # PREDICT
    pred = input_data * weight

    # COMPARE
    error = (pred - goal_pred) ** 2

    # LEARN
    delta = pred - goal_pred
    weight_delta = input_data * delta

    # Mengalikan weight_delta dengan alpha
    weight -= alpha * weight_delta

    print(f"Iterasi {iteration+1} - Error: {error:.6f} | Prediksi: {pred:.6f}")
```

Penjelasan Teoretis:Alpha ($\alpha$): Hyperparameter yang mengatur seberapa cepat model memperbarui bobotnya.Learning Rate terlalu tinggi: Menyebabkan pembaruan bobot melompati titik optimal (overshooting).Learning Rate terlalu rendah: Membutuhkan waktu pelatihan yang sangat lama untuk konvergen.
