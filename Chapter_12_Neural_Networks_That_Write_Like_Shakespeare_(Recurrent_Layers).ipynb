Overview & Teori DasarBab ini mempelajari Recurrent Neural Networks (RNN) yang dirancang untuk mengolah data sekuensial (seperti teks berurutan):   Hidden State (Memori Jaringan): Vektor yang menyimpan informasi dari time step sebelumnya untuk diteruskan ke time step berikutnya.Unrolling over Time: Proses mengesekusi jaringan saraf secara berurutan langkah demi langkah (time step $t$).Weight Sharing Across Time: Menggunakan matriks bobot berulang yang sama di setiap langkah waktu untuk menangani data berpanjang variabel.

1. Arsitektur Recurrent Layer Sederhana (Forward Pass)$$h_t = \text{tanh}(x_t W_{ih} + h_{t-1} W_{hh})$$

```python
import numpy as np

# Set seed agar hasil konsisten
np.random.seed(1)

# Inisialisasi Dimensi
input_dim = 4    # Ukuran alfabet/vocab
hidden_dim = 3   # Ukuran memori hidden state
output_dim = 4   # Ukuran prediksi output

# Bobot Jaringan Recurrent
W_ih = 2 * np.random.random((input_dim, hidden_dim)) - 1  # Input -> Hidden
W_hh = 2 * np.random.random((hidden_dim, hidden_dim)) - 1 # Hidden(t-1) -> Hidden(t)
W_ho = 2 * np.random.random((hidden_dim, output_dim)) - 1 # Hidden -> Output

# Sekuens Input (3 langkah waktu / time steps)
inputs = np.array([
    [1, 0, 0, 0], # t = 0
    [0, 1, 0, 0], # t = 1
    [0, 0, 1, 0]  # t = 2
])

# Hidden state awal (t = -1)
hidden_state = np.zeros((1, hidden_dim))

# Memproses data sekuensial melalui waktu (Unrolling over time)
for t, x in enumerate(inputs):
    x_t = x.reshape(1, input_dim)

    # Menghitung hidden state baru dengan memadukan input saat ini & hidden state sebelumnya
    hidden_state = np.tanh(np.dot(x_t, W_ih) + np.dot(hidden_state, W_hh))

    # Menghitung output pada time step t
    output_t = np.dot(hidden_state, W_ho)

    print(f"Time Step {t} | Hidden State: {hidden_state.round(3)} | Output: {output_t.round(3)}")
```

Penjelasan Teoretis:Recurrent Connection ($W_{hh}$): Menghubungkan memori dari waktu sebelumnya ($h_{t-1}$) ke waktu saat ini ($h_t$).Context Preservation: Memungkinkan jaringan mengingat konteks kata-kata sebelumnya saat memprediksi karakter atau kata berikutnya.
