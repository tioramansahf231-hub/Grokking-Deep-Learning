Overview & Teori DasarBab ini membahas konsep Curriculum Learning dan teknik optimasi tingkat lanjut untuk mempercepat proses konvergensi serta meningkatkan stabilitas pelatihan neural network:Curriculum Learning: Strategi pelatihan yang memberikan contoh data sederhana terlebih dahulu sebelum secara bertahap meningkatkan kompleksitas atau kesulitan data.   Learning Rate Decay (Scheduling): Mengurangi nilai learning rate ($\alpha$) secara berkala seiring berjalannya iterasi agar model tidak melompati titik minimum lokal saat mendekati konvergensi.   SGD dengan Momentum: Menambahkan faktor percepatan (velocity) berdasarkan gradien sebelumnya untuk membantu model melintasi area permukaan rugi (loss surface) yang datar atau bergelombang.

1. Implementasi Learning Rate Decay

```python
import numpy as np

# Inisialisasi parameter
weight = 0.5
goal_pred = 0.8
input_data = 2.0

# Learning Rate Awal dan Decay Factor
alpha_initial = 0.5
decay_rate = 0.05

for iteration in range(20):
    # Dynamic Learning Rate Decay
    alpha = alpha_initial / (1 + decay_rate * iteration)

    # Forward pass
    pred = input_data * weight
    error = (pred - goal_pred) ** 2

    # Backpropagation
    delta = pred - goal_pred
    weight_delta = input_data * delta

    # Update bobot
    weight -= alpha * weight_delta

    print(f"Iterasi {iteration+1:2d} | Alpha: {alpha:.4f} | Error: {error:.6f} | Prediksi: {pred:.4f}")
```

Penjelasan Teoretis:

Adaptasi Learning Rate: Pada awal pelatihan, learning rate bernilai besar agar bobot berpindah cepat menuju solusi. Seiring bertambahnya iterasi, learning rate mengecil untuk penghalusan (fine-tuning) penyesuaian bobot

2. Gradient Descent dengan Momentum$$\text{velocity} = \beta \times \text{velocity} + \alpha \times \text{weight_delta}$$$$\text{weight} = \text{weight} - \text{velocity}$$

```python
import numpy as np

weight = 0.5
goal_pred = 0.8
input_data = 2.0
alpha = 0.05
beta = 0.9  # Momentum factor

velocity = 0

for iteration in range(20):
    pred = input_data * weight
    error = (pred - goal_pred) ** 2

    delta = pred - goal_pred
    weight_delta = input_data * delta

    # Menghitung Velocity Momentum
    velocity = beta * velocity + alpha * weight_delta
    weight -= velocity

    print(f"Iterasi {iteration+1:2d} | Velocity: {velocity:.6f} | Error: {error:.6f}")
```

Penjelasan Teoretis:Momentum ($\beta$): Mengakumulasi gradien masa lalu untuk mempercepat pergerakan bobot ke arah yang konsisten dan meredam osilasi
